## 1. Create an interactive bar chart showing total quantity and revenue by country (excluding United Kingdom) for the month of April 2011.

```python
# Converting InvoiceDate to datetime format 
data['InvoiceDate'] = pd.to_datetime(data['InvoiceDate'])

# Extracting Year and Month
data['Year'] = data['InvoiceDate'].dt.year
data['Month'] = data['InvoiceDate'].dt.month

april_data = data[(data['Year'] == 2011) & (data['Month'] == 4) & (data['Country'] != 'United Kingdom')]

# Grouping data by country
country_summary = april_data.groupby('Country').agg({'Quantity': 'sum', 'Revenue': 'sum'}).reset_index()

#Interactive bar chart
country_summary.iplot(kind='bar', x='Country', y=['Quantity', 'Revenue'], title='April 2011: Total Quantity & Revenue by Country', xTitle='Country')
```

## 2. Create an interactive line chart showing quantity and revenue sold to France between January 1st and May 31st 2011.

```python
#Filtering France within the data
france_data = data[(data['Year'] == 2011) & (data['Month'].between(1, 5)) & (data['Country'] == 'France')]

# Grouping by date
france_summary = france_data.groupby('InvoiceDate').agg({'Quantity': 'sum', 'Revenue': 'sum'}).reset_index()

france_summary['InvoiceDate'] = france_summary['InvoiceDate'].dt.strftime('%Y-%m-%d')

#Interactive bar chart
france_summary.iplot(kind='line', x='InvoiceDate', y=['Quantity', 'Revenue'], title='Quantity & Revenue Sold to France (Jan-May 2011)', xTitle='Date')
```

## 3. Create an interactive scatter plot showing the relationship between average quantity (x-axis) and average unit price (y-axis) for the product PARTY BUNTING with the plot points color-coded by country (categories).

```python
# Filtering for "PARTY BUNTING"
party_bunting_data = data[data['Description'] == 'PARTY BUNTING']

# Grouping by country
party_summary = party_bunting_data.groupby('Country').agg({'Quantity': 'mean', 'UnitPrice': 'mean'}).reset_index()

# Interactive scatter plot
fig = px.scatter(party_summary, x='Quantity', y='UnitPrice', color='Country',
                 title='Average Quantity vs. Average Unit Price for PARTY BUNTING',
                 labels={'Quantity': 'Avg Quantity', 'UnitPrice': 'Avg Unit Price'},
                 hover_data=['Country'])

fig.show()
```

## 4. Create a set of interactive histograms showing the distributions of quantity per invoice for the following countries: EIRE, Germany, France, and Netherlands.

```python
#Filtering EIRE,Germany, Frnace and Netherlands
selected_countries = ['EIRE', 'Germany', 'France', 'Netherlands']
filtered_data = data[data['Country'].isin(selected_countries)]

#Grouping by InvoiceNo and calculating the total quantity for each invoice
grouped_data = filtered_data.groupby(['Country', 'InvoiceNo'])['Quantity'].sum().reset_index()

#Interactive Histograms
fig = px.histogram(grouped_data, x='Quantity', nbins=30, color='Country',
                   facet_col='Country',  
                   title='Quantity Per Invoice by Country',
                   labels={'Quantity': 'Total Quantity per Invoice'})

fig.show()
```

## 5. Create an interactive side-by-side bar chart showing the revenue by country listed below (bars) for each of the products listed below.

```python
product_list = ['JUMBO BAG RED RETROSPOT', 
                'CREAM HANGING HEART T-LIGHT HOLDER',
                'REGENCY CAKESTAND 3 TIER']

country_list = ['EIRE', 'Germany', 'France', 'Netherlands']
```

## 6. Create an interactive line chart showing quantity sold by day for the United Kingdom. Add drop-down boxes for Year and Month that allow you to filter the date range that appears in the chart.

```python
data['Year'] = pd.DatetimeIndex(data['InvoiceDate']).year
data['Month'] = pd.DatetimeIndex(data['InvoiceDate']).month
data['Day'] = pd.DatetimeIndex(data['InvoiceDate']).day
uk = data[data['Country']=='United Kingdom']
```

```python
#Aggregating Quantity Sold Per Day
daily_sales = uk.groupby(['Year', 'Month', 'Day'])['Quantity'].sum().reset_index()
daily_sales['Date'] = pd.to_datetime(daily_sales[['Year', 'Month', 'Day']])
daily_sales['Date'] = daily_sales['Date'].dt.strftime('%Y-%m-%d')

#Interactive Line Chart
def plot_sales(selected_year, selected_month):
    filtered_data = daily_sales[(daily_sales['Year'] == selected_year) & (daily_sales['Month'] == selected_month)]
    filtered_data.iplot(kind='line', x='Date', y='Quantity', 
                        title=f'Quantity Sold Per Day in UK ({selected_year}-{selected_month})', 
                        xTitle='Date', yTitle='Quantity')

interact(plot_sales, selected_year=daily_sales['Year'].unique(), selected_month=daily_sales['Month'].unique())
```

## 7. Create an interactive scatter plot that plots number of invoices (x-axis) vs. number of customers (y-axis) and the plot points represent individual products. Add two sliders that control the x and y axis ranges.

```python
agg_func = {'InvoiceNo':'nunique',
            'Quantity':'sum',
            'UnitPrice':'mean',
            'Revenue':'sum',
            'CustomerID':'nunique'}

products = uk.groupby('Description').agg(agg_func).reset_index()
```

```python
#Interactive scatter plot
def plot_scatter(x_range, y_range):
    filtered_data = products[(products['InvoiceNo'] <= x_range) & 
                                     (products['CustomerID'] <= y_range)]

    fig = px.scatter(filtered_data, x='InvoiceNo', y='CustomerID', 
                     hover_data=['Description'], 
                     title=f'Invoices vs. Customers (Max X: {x_range}, Max Y: {y_range})',
                     labels={'InvoiceNo': 'Unique Invoice Count', 'CustomerID': 'Unique Customer Count'})

    fig.show()


interact(plot_scatter, x_range=(1, products['InvoiceNo'].max()), 
                      y_range=(1, products['CustomerID'].max()))
```

## 8. Creat an interactive bar chart that shows revenue by product description. Add a text field widget that filters the results to show the product that contain the text entered in their description.

```python
# Group data by product description and calculate total revenue
product_revenue = data.groupby('Description')['Revenue'].sum().reset_index()
def plot_filtered_bar_chart(search_text):
    # Filter descriptions based on user input 
    filtered_data = product_revenue[product_revenue['Description'].str.contains(search_text, case=False, na=False)]

    fig = px.bar(filtered_data, x='Description', y='Revenue', 
                 title=f'Revenue by Product (Filtered: "{search_text}")',
                 labels={'Description': 'Product Description', 'Revenue': 'Total Revenue'},
                 height=600)

    fig.show()


text_widget = widgets.Text(
    placeholder='Enter product keyword...',
    description='Search:',
    continuous_update=False
)

interact(plot_filtered_bar_chart, search_text=text_widget)
```
